```python
def int_to_roman(num):
    val = [1000, 900, 500, 400,100, 90, 50, 40,10, 9, 5, 4, 1]
    syms = ["M", "CM", "D", "CD","C", "XC", "L", "XL","X", "IX", "V", "IV", "I" ]
    roman_num = ""
    i = 0
    while num > 0:
        for _ in range(num // val[i]):
            print(val[i])
            roman_num = roman_num + syms[i]
            print(f"num: {num}, val: {val[i]}, roman_num: {roman_num}")
            num = num - val[i]
            print(i)            
        i += 1
    return roman_num

# Example usage:
number = 15
roman_number = int_to_roman(number)
display(roman_number)

```
